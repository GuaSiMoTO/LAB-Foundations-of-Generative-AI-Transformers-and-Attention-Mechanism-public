## Iteration 1: Loading the Models
1. **BART**:

2. **Sentiment Analysis Model**:

3. **Translation Model**:
   - Import and initialize the pre-trained **Helsinki-NLP/opus-mt-en-es** model.

````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

# Load BART
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Load DistilBERT for sentiment analysis
distilbert_model = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")


# Load Helsinki-NLP
translator = pipeline("translation_en_to_es", model="Helsinki-NLP/opus-mt-en-es")
````

## Iteration 2: Text Summarization with BART
1. Choose a text file (`article.txt`) containing substantial content for summarization.
2. Use **BART** to generate a summary:
   - Define a function that takes the input text and returns a summary using **BART**.
   - Save the summary to `summary_bart.txt`.

````python
# Load the text from the file
with open("article.txt", "r") as file:
    article_text = file.read()

# Generate summary using the function
def generate_bart_summary(input_text, max_length=130, min_length=30):
    summary = summarizer(input_text, max_length=max_length, min_length=min_length, do_sample=False)
    return summary[0]['summary_text']

bart_summary = generate_bart_summary(article_text)

# Save summary
with open("summary_bart.txt", "w") as file:
    file.write(bart_summary)
````

## Iteration 3: Translating Text with Helsinki-NLP
1. **Generate a Translation**:
   - Use the tokenized input to translate the text from English to Spanish using the Helsinki-NLP model.
   - Experiment with parameters like `max_length` and `num_beams` to explore how they affect the translation output.

2. **Save Translations**:
   - Save the translations to a file `translations_helsinki.txt`.

````python
# Translate the article to Spanish

def translate_text_to_spanish(text, max_length=500, num_beams=4):
    translation = translator(
        text,
        max_length=max_length,
        num_beams=num_beams,
        early_stopping=True
    )
    return translation[0]['translation_text']

# Uso de la función
with open("article.txt", "r") as file:
    summary_text = file.read()

translated_text = translate_text_to_spanish(summary_text)

# Save the translation to a file
with open('translations_helsinki.txt', 'w') as file:
    file.write(translated_text)
````

## Iteration 4: Sentiment Analysis with DistilBERT or RoBERTa
1. Prepare a set of short texts with sentiment labels (positive, negative).
2. Use the pretrained model (e.g., **DistilBERT**) for sentiment classification:
   - Define a function to process each text and return a sentiment label.
   - Save the predictions to `sentiments_model.txt`.

````python
# Example texts
texts = [
    "I love the new design of the product!",  
    "The service was terrible and disappointing.",  
    "It's okay, but nothing special.", 
    "This is absolutely amazing! Best purchase ever!",  
    "I've never been so frustrated with a product in my life.", 
    "The item works as described, no complaints.",  
    "The app crashes constantly and loses all my data."
]

# Analyze sentiments

def analyze_sentiments(text_list, model):
    return model(text_list)

# Uso de la función
results = analyze_sentiments(texts, distilbert_model)

with open("sentiments_model.txt", "w") as file:
    for text, res in zip(texts, results):
        file.write(f"Text: {text}\nSentiment: {res['label']}, Confidence: {res['score']:.4f}\n\n")
````
