## Iteration 2: Generate Text and Extract Attention Weights

Next, we will define a function to generate text and extract the attention weights from both models. 


````python
# Function to generate text and extract attention weights
def generate_text_with_attention(model, tokenizer, prompt, max_length=50):
    input_ids = tokenizer.encode(prompt, return_tensors='tf')
    outputs = model(input_ids, output_attentions=True)
    attentions = outputs.attentions  # List of attention weights for all layers
    generated_text = model.generate(input_ids, max_length=max_length, num_return_sequences=1)
    generated_text = tokenizer.decode(generated_text[0], skip_special_tokens=True)
    return generated_text, attentions
````

We will now visualize the attention matrices to understand how the models attend to different parts of the input.

We'll define a function to plot the attention matrices and then visualize them for both models.

````python
# Show an averaged visualization of the attention heads from the last layer, providing a more global perspective.
def plot_attention_weights(attentions, title):
    attention_weights = attentions[-1]  # Última capa
    attention_weights = tf.reduce_mean(attention_weights, axis=1)  # Promediar sobre las cabezas de atención
    plt.figure(figsize=(10, 8))
    sns.heatmap(attention_weights[0], cmap='viridis')
    plt.title(title)
    plt.xlabel('Tokens')
    plt.ylabel('Tokens')
    plt.show()

````
